```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().resolve().parents[1]))
from paths import MODELS

from huggingface_hub import snapshot_download

models = {
    "facebook/pe-av-small": str(MODELS / "manual/facebook--pe-av-small"),
    "facebook/pe-av-base": str(MODELS / "manual/facebook--pe-av-base"),
    "facebook/pe-av-base-16-frame": str(MODELS / "manual/facebook--pe-av-base-16-frame"),
    "facebook/pe-av-large-16-frame": str(MODELS / "manual/facebook--pe-av-large-16-frame"),
}

for repo_id, local_dir in models.items():
    path = snapshot_download(
        repo_id=repo_id,
        local_dir=local_dir,
        local_dir_use_symlinks=False,
        resume_download=True,
    )
    print(f"{repo_id} downloaded to: {path}")
```
